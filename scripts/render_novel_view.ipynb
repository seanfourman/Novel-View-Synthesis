# Render novel-view asset for slide 2

Produces `assets/images/redtoyota_novel_view.jpg` - the clean inpainted novel view shown at the final "refinement" stage of the depth-based NVS slide.

Runs Depth Anything V2 Small + forward warp with the exact same parameters as the in-browser pipeline, then LaMa for hole filling.

**How to use in Colab:**
1. Upload `assets/images/redtoyota.jpg` and `scripts/render_novel_view.py` to the Colab workspace (top of left sidebar → upload).
2. Run the cells below in order.
3. Download `redtoyota_novel_view.jpg` from the Colab file browser and drop it into `assets/images/` in this repo.

```python
# Make the layout match the local repo so render_novel_view.py finds the image.
import os, shutil, pathlib
pathlib.Path('assets/images').mkdir(parents=True, exist_ok=True)
if os.path.exists('redtoyota.jpg') and not os.path.exists('assets/images/redtoyota.jpg'):
    shutil.move('redtoyota.jpg', 'assets/images/redtoyota.jpg')
pathlib.Path('scripts').mkdir(exist_ok=True)
if os.path.exists('render_novel_view.py') and not os.path.exists('scripts/render_novel_view.py'):
    shutil.move('render_novel_view.py', 'scripts/render_novel_view.py')
print(os.listdir('assets/images'))
```
